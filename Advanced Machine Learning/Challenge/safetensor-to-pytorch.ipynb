```python
import torch
from safetensors.torch import load_file
import os

def convert_safetensors_to_pth(input_file, output_file):
    print(f"Caricamento modello: {input_file}")

    # Carica i tensori
    state_dict = load_file(input_file, device="cpu")

    print(f"Parametri caricati: {len(state_dict)}")

    # Salva in formato PTH
    torch.save(state_dict, output_file)

    print(f"\nConversione completata!")
    print(f"File salvato: {output_file}")
    print(f"Dimensione: {os.path.getsize(output_file) / (1024**3):.2f} GB")
```

```python
input_file  = "weights/dinov3-vitb16-pretrain-lvd1689m.safetensors"
output_file = "weights/dinov3-vitb16-pretrain-lvd1689m.pth"

convert_safetensors_to_pth(input_file, output_file)
```
